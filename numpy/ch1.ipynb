```python
def pySun ():
    a=[1,8,9,13,14]
    b=[9,7,5,15,19]
    C=[7,5,4,23,15]
    d=[]
    for i in range(len(a)):
        d.append(a[i]**2+b[i]**2+C[i]**3)
    return d
%time print(pySun())
```

```python
def npSun():
    a=np.array([1,8,9,13,14])
    b=np.array([9,7,5,15,19])
    c=np.array([7,5,4,23,15])
    d=a**2+b**2+c**3
    return d
%time print(npSun())
```

```python
#导入numpy库，取名np
import numpy as np
#默认d0=1,产生[0，1)范围内的一个浮点随机数
a= np.random.rand()
#d0=2，产生[0，1)范围内的一维数组的浮点随机数
b= np.random.rand(2)
#d0=2,d1=3,产生[0,1)范围内的二维数组的浮点随机数
c = np.random.rand(2,3)
print(a)
print(b)
print(c)
```
