## 5.3 正则表达式

### 常用正则表达式符号
- `.`: 匹配任意字符 (除了换行符)。
- `*`: 匹配前一个字符 0 次或多次。
- `+`: 匹配前一个字符 1 次或多次。
- `?`: 匹配前一个字符 0 次或 1 次。
- `^`: 匹配字符串开头。
- `$`: 匹配字符串结尾。
- `[]`: 匹配字符集合，如 `[a-z]`。
- `\d`: 匹配数字。
- `\w`: 匹配字母、数字、下划线。
- `\s`: 匹配空白字符。

```python
#待匹配的字符串
title ="你好,hello,世界"
#创建正则表达式，用于只匹配中文
pattern=re.compile(r"[\u4e00-\u9fa5]+")
#检索整个字符串，将匹配的中文放到列表中
result=pattern.findall(title)
print(result)
```

### 正则表达式例子

```python
def displaymatch(match):
    if match is None:
        return None
    return '<Match: %r, groups=%r>' % (match.group(), match.groups())
#group() 方法用于获取匹配的字符串，groups() 方法用于获取所有捕获的分组。
```

假设你在写一个扑克程序，一个玩家的一手牌为五个字符的串，每个字符表示一张牌，"a" 就是 A, "k" K， "q" Q, "j" J, "t" 为 10, "2" 到 "9" 表示2 到 9。
要看给定的字符串是否有效，我们可以按照以下步骤

```python
valid = re.compile(r"^[a2-9tjqk]{5}$")
#compile() 函数用于编译正则表达式，生成一个正则表达式（ Pattern ）对象，供 match() 和 search() 这两个函数使用。
# ^ 表示字符串的开头，$ 表示字符串的结尾，{5} 表示前面的字符必须出现5次。
# 因此，整个正则表达式 ^[a2-9tjqk]{5}$ 的意思是：字符串必须完全由5个字符组成，这些字符只能是 a, 2-9, t, j, q, k 中的任意一个。
displaymatch(valid.match("akt5q"))  # 有效。
```

```python
displaymatch(valid.match("akt5e"))  # 无效。
```

```python
displaymatch(valid.match("akt"))    # 无效。
```

```python
displaymatch(valid.match("727ak"))  # 有效。
#没有分组是因为正则表达式中没有使用括号 () 来定义捕获组，所以 groups() 方法返回一个空的元组 ()。
```

最后一手牌，"727ak" ，包含了一个对子，或者两张同样数值的牌。要用正则表达式匹配它，应该使用向后引用如下

```python
pair = re.compile(r".*(.).*\1")
# .* 表示任意字符出现任意次，(.) 表示匹配任意单个字符并捕获该字符，\1 表示引用第一个捕获的字符。
#引用的意思是，正则表达式在匹配时，\1 会被替换为第一个捕获的字符。
# 因此，整个正则表达式 .*(.).*\1 的意思是：字符串中有任意字符，后面跟着一个字符（捕获），再后面跟着任意字符，最后再跟着与前面捕获的字符相同的字符。






displaymatch(pair.match("377aa"))    # 成对的 7 和 a。
```

```python
displaymatch(pair.match("717ak"))     # 成对的 7。
```

```python
displaymatch(pair.match("718ak"))     # 没有成对。
```

```python
displaymatch(pair.match("354aa"))     # 成对的 a。
```

```python
displaymatch(pair.match("377aa"))    # 成对的 7 和 a。
# 因为正则表达式只捕获了第一个成对的字符，所以 groups() 方法返回的元组中只有一个元素，即第一个成对的字符
# 如果想要返回两个成对的字符，需要修改正则表达式，增加一个捕获组，例如：
```

要找出对子由什么牌组成，开发者可以按照下面的方式来使用匹配对象的 group() 方法:

```python
pair = re.compile(r".*(.).*\1")
# .* 表示任意字符出现任意次，(.) 表示匹配任意单个字符并捕获该字符，\1 表示引用第一个捕获的字符。




```

```python
pair.match("717ak").group(1)
```

### 文字整理

sub() 替换字符串中出现的样式的每一个实例。这个例子证明了使用 sub() 来整理文字，或者随机化每个字符的位置，除了首位和末尾字符

```python
def repl(m):
    inner_word = list(m.group(2))
    random.shuffle(inner_word)
    return m.group(1) + "".join(inner_word) + m.group(3)

text = "Professor Abdolmalek, please report your absences promptly."
re.sub(r"(\w)(\w+)(\w)", repl, text)

re.sub(r"(\w)(\w+)(\w)", repl, text)
```

### lxml 常用方法
- `etree.Element(tag)`: 创建节点。
- `etree.tostring(element)`: 将节点转换为字符串。
- `etree.fromstring(xml_str)`: 从字符串解析 XML。
- `etree.parse(file_path)`: 从文件解析 XML。
- `element.xpath(path)`: 使用 XPath 查找节点。

```python
root=etree.Element('root')
```

（1）节点操作：用tag属性获取节点名称

```python
print(root.tag)
```

（2）节点属性操作：创建节点同时可以为节点增加属性

```python
#创建root节点，并为其添加属性
root=etree.Element('root',interesting='totally')
print(etree.tostring(root))
```

可以通过set()方法为已有的节点添加属性。

```python
#再次给root节点添加age属性
root.set('age','30')
print(etree.tostring(root))
```

(3)节点内文本的操作：可以通过text、tail属性或者Xpath()方法来访问文件内容

```python
root=etree.Element('root')
root.text='hello world'
print(root.text)
print(etree.tostring(root))
```

从字符串或文件中解析XML

```python
xml_data = '<root>data</root>'
#fromstring()方法
root_one = etree.fromstring(xml_data)
print(root_one.tag)
print(etree.tostring(root_one))
```

```python
#XML方法,与fromstring方法基本一样
root_two=etree.XML(xml_data)
print(root_two.tag)
print(etree.tostring(root_two))
```

```python
#HTML()方法，如果没有<html>标签和<body>标签，会自动补上
root_three=etree.HTML(xml_data)
print(root_three.tag)
print(etree.tostring(root_three))
```

还可调用parse()函数从XML文件中直接解析

```python


html=etree.parse('./hello.html')

result=etree.tostring(html,pretty_print=True)
```

3.ElementPath类简介

```python
#从字符串中解析XML，返回根节点
root=etree.XML("<root><a x='123'>aText<b/><c/><b/></a></root>")
#从根节点中查找，返回匹配到的节点名称
print(root.find("a").tag)
#从根节点开始查找，返回匹配到的第一个节点的名称
print(root.findall(".//a[@x]")[0].tag)
```

### 5.4.5 lxml库的基本使用

```python
html=etree.parse('hello.html')
#查找所有的li节点
result=html.xpath('//li')
#打印<li>标签的元素集合
print(result)
#打印<li>标签的个数
print(len(result))
#打印返回结果的类型
print(type(result))
#打印第一个元素的类型
print(type(result[0]))
```

继续获取<li>标签的class属性

```python
html=etree.parse('hello.html')
#查找所有li节点的class属性
result=html.xpath('//li/@class')
print(result)
```

获取倒数第二个元素的内容

```python
html=etree.parse('hello.html')
#获取倒数第二个元素的内容
result=html.xpath('//li[last()-1]/a')
print(result[0].text)
```

### 5.5.2 构建BeautifulSoup对象

 def_init_(self,markup="",features=None,builder=None,parse_only=None,from_encoding=None,exclude_encodings=None,**kwargs)  
上述方法的参数说明  
markup:要解析的HTML或XML文档字符串。  
features:指定解析器的类型，如"html.parser"、"lxml"、"xml"等。  
builder:自定义的解析器构建器。  
parse_only:用于过滤特定标签的解析器。  
from_encoding:指定输入字符串的编码格式。  
exclude_encodings:指定不使用的编码格式列表。  
kwargs:其他可选参数，用于传递给解析器的额外配置选项。  

```python
html_doc = """
<html><head><title>The Dormouse's story</title></head>
<body>
<p class="title"><b>The Dormouse's story</b></p>

<p class="story">Once upon a time there were three little sisters; and their names were
<a href="http://example.com/elsie" class="sister" id="link1">Elsie</a>,
<a href="http://example.com/lacie" class="sister" id="link2">Lacie</a> and
<a href="http://example.com/tillie" class="sister" id="link3">Tillie</a>;
and they lived at the bottom of a well.</p>

<p class="story">...</p>
"""
soup=BeautifulSoup(html_doc, 'lxml')
print(soup.prettify())
```

### JSONPath 常用语法
- `$`: 根节点。
- `.`: 子节点。
- `..`: 递归下降 (任意位置)。
- `*`: 通配符。
- `[]`: 下标操作符。
- `[start:end]`: 数组切片。
- `?()`: 过滤表达式。
- `@`: 当前节点。

```python
books=json.load(open("bookJson.json", encoding='utf-8'))
checkurl="$.store.bicycle.color"
print(jsonpath.jsonpath(books, checkurl))
```

```python
checkurl="$.store.book[*]"
object_list=jsonpath.jsonpath(books, checkurl)
print(object_list)
```

```python
checkurl="$.store.book[0]"
obj=jsonpath.jsonpath(books, checkurl)
print(obj)
```

```python
checkurl="$.store.book[*].title"
titles=jsonpath.jsonpath(books,checkurl)
print(titles)
```

```python
checkurl="$.store.book[?(@.category=='fiction')]"
titles=jsonpath.jsonpath(books,checkurl)
print(books)
```

```python
checkurl="$.store.book[?(@.price<10)]"
books=jsonpath.jsonpath(books, checkurl)
print(books)
```

```python
books=json.load(open("bookJson.json", encoding='utf-8'))
checkurl="$.store.book[?(@.isbn)]"
books=jsonpath.jsonpath(books, checkurl)
print(books)
```

```python
#5.6.7案例——获取拉钩网城市列表
import urllib.request
import jsonpath
import json
```

```python
url='http://www.lagou.com/lbs/getAllCitySearchLabels.json'
request=urllib.request.Request(url)
response=urllib.request.urlopen(request)
html=response.read()
print(html)
```

```python
#把Json格式字符串转换成python对象
jsonobj=json.loads(html)
#从根节点开始，匹配name节点
city_list=jsonpath.jsonpath(jsonobj,'$..name')
print(city_list)
```

```python
#打开或创建一个名为city,json的文件
file=open('city.json','w')
#将列表序列化为json格式的字符串
content=json.dumps(city_list,ensure_ascii=False)
#写入本地文件
file.write(content)
file.close()
```
