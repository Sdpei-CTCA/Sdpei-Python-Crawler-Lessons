## 11. 注释与数据可视化案例
使用 `ax.annotate()` 方法在图表中添加文本注释，常用于标记重要数据点。

```python
import pandas as pd
from datetime import datetime
fig,ax= plt.subplots()
data = pd.read_csv("spx.csv", index_col=0, parse_dates=True)
spx = data["SPX"]
spx.plot(ax=ax, color="black", figsize=(10, 5))
crisis_data=[
    (datetime(2007, 10, 11), "Peak of bull market"),
    (datetime(2008, 3, 12), "Bear Stearns Fails"),
    (datetime(2008, 9, 15), "Lehman Bankruptcy")
]
for date, label in crisis_data:
    ax.annotate(label, xy=(date, spx.asof(date) + 75),
            xytext=(date, spx.asof(date) + 225),
            arrowprops=dict(facecolor="black",headwidth=4, width=2,headlength=4),
            horizontalalignment="left",verticalalignment="top")
# Zoom in on 2007-2010
ax.set_xlim(["1/1/2007","1/1/2011"])
ax.set_ylim([600,1800])
ax.set_title("Important dates in the 2008-2009 financial crisis")
```

```python
fig,ax= plt.subplots()
data = pd.read_csv("spx.csv", index_col=0, parse_dates=True)
spx = data["SPX"]
spx.plot(ax=ax, color="black", figsize=(10, 5))
crisis_data=[
(datetime(2007, 10, 11), "Peak of bull market"),
(datetime(2008, 3, 12), "Bear Stearns Fails"),
(datetime(2008, 9, 15), "Lehman Bankruptcy")
]
for date, label in crisis_data:
    ax.annotate(label, xy=(date, spx.asof(date) + 75),
                xytext=(date, spx.asof(date) + 225),
                arrowprops=dict(facecolor="yellow", headwidth=6, width=3,
                                headlength=4),
                horizontalalignment="left", verticalalignment="top")
# Zoom in on 2007-2010
ax.set_xlim(["1/1/2007","1/1/2011"])
ax.set_ylim([600,1800])
ax.set_title("Important dates in the 2008- 2009 financial crisis")
```

## 12. 绘制图形对象 (Patches)
可以使用 `matplotlib.patches` 中的类（如 `Rectangle`, `Circle`, `Polygon`）创建图形对象，并使用 `ax.add_patch()` 添加到子图中。

```python
fig,ax = plt.subplots()
rect = plt.Rectangle((0.2, 0.75), 0.4, 0.15, color="black", alpha=0.3)
circ = plt.Circle((0.7, 0.2), 0.15, color="blue", alpha=0.3)
pgon= plt.Polygon([[0.15,0.15], [0.35,0.4], [0.2, 0.6]],
color="green", alpha=0.5)
ax.add_patch(rect)
ax.add_patch(circ)
ax.add_patch(pgon)
```

## 3. Pyecharts 交互式图表
`pyecharts` 是一个用于生成 Echarts 图表的 Python 库。Echarts 是百度开源的一个数据可视化 JS 库，生成的图表具有良好的交互性。

**基本使用步骤**：
1. 导入图表类（如 `Bar`, `Line`）和配置项 (`options`)。
2. 创建图表对象。
3. 添加数据 (`add_xaxis`, `add_yaxis`)。
4. 设置全局配置 (`set_global_opts`)。
5. 渲染图表 (`render_notebook()` 在 Jupyter 中显示)。

```python
from pyecharts.charts import Bar
from pyecharts import options as opts

# 1. 创建柱状图对象
bar = Bar()

# 2. 添加 X 轴数据
bar.add_xaxis(["衬衫", "羊毛衫", "雪纺衫", "裤子", "高跟鞋", "袜子"])

# 3. 添加 Y 轴数据
bar.add_yaxis("商家A", [5, 20, 36, 10, 75, 90])
bar.add_yaxis("商家B", [15, 6, 45, 20, 35, 66])

# 4. 设置全局配置项 (标题、工具箱等)
bar.set_global_opts(
    title_opts=opts.TitleOpts(title="某商场销售情况", subtitle="Pyecharts 示例"),
    toolbox_opts=opts.ToolboxOpts(), # 显示工具箱
    legend_opts=opts.LegendOpts(is_show=True) # 显示图例
)

# 5. 在 Notebook 中渲染
bar.render_notebook()
```

## 散点图

```python
macro= pd.read_csv("macrodata.csv")
data= macro[["cpi","m1","tbilrate","unemp"]]
trans_data=np.log(data).diff().dropna()
trans_data.tail()
```

使用 `sns.regplot` 绘制散点图并自动添加线性回归拟合线，用于观察两个变量之间的关系。

```python
ax = sns.regplot(x="m1", y="unemp", data=trans_data)
ax.set_title("Changes in log(ml) versus log(unemp)")
```

使用 `sns.pairplot` 可以一次性绘制数据集中所有数值变量两两之间的散点图矩阵，对角线上显示单变量分布。

```python
sns.pairplot(trans_data, diag_kind="kde", plot_kws={"alpha":0.2})
```
