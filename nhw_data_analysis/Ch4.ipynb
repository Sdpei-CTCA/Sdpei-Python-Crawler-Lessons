## 4.5 代理服务器

### 4.5.1 简单的自定义opener

```python
#构建一个HTTPHandler 处理器对象，支持处理HTTP请求
#如果在HTTPHandler()方法中增加参数debuglevel=1，会将Debug log打开，自动将收包和发包的报头打印出来方便调试
http_handler = urllib.request.HTTPHandler(debuglevel=1)
#调用urllib2.build_opener()方法，创建支持处理HTTP请求的opener对象
opener=urllib.request.build_opener(http_handler)
#构建Request请求，添加User-Agent伪装成浏览器
headers = {
	"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/114.0.0.0 Safari/537.36"
}
request = urllib.request.Request("https://www.bilibili.com/", headers=headers)
#调用自定义opener对象的open方法()，发送request请求
#注意区别：不再通过urllib.request.urlopen()发送请求
response = opener.open(request)
#获取服务器响应内容
print(response.read())
```

### 4.5.2 设置代理服务器

```python
httpproxy_handler=urllib.request.ProxyHandler({"http":"124.88.67.81:80"})
nullproxy_handler=urllib.request.ProxyHandler({})
proxy_switch=True
#通过urllib.request.build_opener方法使用代理handler对象创建自定义opener对象
#根据代理开关是否打开，使用不同的代理模式
if proxy_switch:
    opener=urllib.request.build_opener(httpproxy_handler)
else:
    opener=urllib.request.build_opener(nullproxy_handler)
request=urllib.request.Request("https://www.bilibili.com/", headers=headers)
response=opener.open(request)
print(response.read())
```

## 4.7 常见的网络异常

### 4.7.1 URLError异常和捕获

URLError 产生的原因主要有以下几种：
（1）没有连接网络
（2）服务器连接失败
（3）找不到指定的服务器
可以使用try...except语句捕获相应的异常。例如：

```python
request=urllib.request.Request("http://www.ajkfhafwjqh.com")
try:
    urllib.request.urlopen(request,timeout=5)
except urllib.error.URLError as err:
    print(err)
```

### 4.7.2 HttpError异常和捕获

```python
request=urllib.request.Request('http://www.itcast.cn/net')
try:
    urllib.request.urlopen(request)
except urllib.error.HTTPError as e:
    print(e.code)
```

## 4.8 更人性化的 requests 库
- `urllib` 是 Python 标准库，功能强大但使用较为繁琐。
- `requests` 是第三方库，基于 `urllib`，但 API 设计更加人性化，使用更简单。
- 下面对比使用 `urllib` 和 `requests` 发送带参数和 Headers 的 GET 请求。

```python
#(1)使用urllib库以GET请求的方式爬取网页
#请求的URL路径和查询参数
url="https://www.baidu.com/"
word={"wd":"山东体育学院"}
#转换成url编码格式(字符串)
word=urllib.parse.urlencode(word)
#拼接完整的URL的路径
new_url=url+'?'+word
#请求报头
headers = {
	"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/114.0.0.0 Safari/537.36"
}
#根据URL和headers构建Request请求对象
request=urllib.request.Request(new_url,headers=headers)
#发送请求
response=urllib.request.urlopen(request)
#使用read()方法读取响应内容
html=response.read().decode('utf-8')
print(html)
```

```python
#使用requests库发送GET请求
#请求的URL路径和查询参数
url="https://www.baidu.com/"
param={"wd":"山东体育学院"}
#请求报头
headers = {
	"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/114.0.0.0 Safari/537.36"
}
#发送GET请求

response=requests.get(url,params=param,headers=headers)
response.encoding=response.apparent_encoding

#获取响应内容
print(response.text)
```

## 案例——爬取 B 站搜索结果
本案例包含三个主要函数：
1.  `bilibili_spider(url, begin_page, end_page)`: 爬虫调度函数，负责循环爬取指定页码范围的页面。
2.  `load_page(url)`: 页面加载函数，负责发送请求并获取响应内容。
3.  `write_page(html, filename)`: 页面保存函数，负责将获取的 HTML 内容写入文件。

第一页：https://tieba.baidu.com/f?kw=%E8%B6%85%E7%BA%A7%E9%A9%AC%E9%87%8C%E5%A5%A5&fr=index
第二页：https://tieba.baidu.com/f?kw=%E8%B6%85%E7%BA%A7%E9%A9%AC%E9%87%8C%E5%A5%A5&ie=utf-8&pn=50

```python
def bilibili_spider(url,begin_page,end_page):
    for page in range(begin_page,end_page+1):
        on=(page-1)*30
        file_name="第"+str(page)+"页.html"
        full_url=url+"&page="+str(page)+"&o="+str(on)
        html=load_page(full_url)
        write_page(html,file_name)
```

```python
def load_page(url):
    
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.3"
    }
    request=urllib.request.Request(url,headers=headers)
    return urllib.request.urlopen(request).read().decode('utf-8')
```

```python
def write_page(html,filename):
    print("正在保存"+filename)
    with open(filename,"w",encoding="utf-8") as f:
        f.write(html)
```

```python
if __name__ == "__main__":
    kw=input("请输入要搜索的关键字：")
    begin_page=int(input("请输入起始页："))
    end_page=int(input("请输入终止页："))
    url="https://search.bilibili.com/all?"
    key=urllib.parse.urlencode({"kw":kw})
    #组合后的url:https://search.bilibili.com/all?kw=关键字
    url=url+key
    bilibili_spider(url,begin_page,end_page)
```
