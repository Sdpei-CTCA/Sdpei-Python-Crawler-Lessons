```python
import urllib.request
import urllib.error
import urllib.parse
import requests
```

## 4.1 urllib库概述  

urllib库是Python内置的HTTP请求库，它可以看作处理URL的组件集合。urllib库包括四大模块  

(1)urllib.request：请求模块  

(2)urllib.error:异常处理模块

(3)urllib.parse：URL解析模块

(4)urllib.robotparser:robot.txt解析模块

## 4.2 快速使用urllib爬取网页

### 4.2.1 快速爬取一个网页

```python
#调用urllib.request库中的urlopen方法，并向其传递一个url。
response=urllib.request.urlopen('https://sdpei-ctca.tech/')
#使用read()方法读取获得到的网页内容
html=response.read().decode('UTF-8')
#打印网页内容
print(html)
```

### 4.2.2 分析urlopen()方法

### 4.2.3 使用HTTPResponse对象

```python
response=urllib.request.urlopen('http://hitwith.top/')
print(type(response))
```

```python
response=urllib.request.urlopen('http://python.org')
#获取响应信息对应的URL
print(response.geturl())
#获取相应码
print(response.getcode())
#获取页面的元信息
print(response.info())
```

## 4.3 使用urllib实现数据传输

### 4.3.1 URL码转换

```python
data={
    'a':'山东体育学院',
    'b':'玩机社'
}
result=urllib.parse.urlencode(data)
print(result)
```

```python
result=urllib.parse.unquote('a=%E5%B1%B1%E4%B8%9C%E4%BD%93%E8%82%B2%E5%AD%A6%E9%99%A2&b=%E7%8E%A9%E6%9C%BA%E7%A4%BE')
print(result)
```
